```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df = pd.read_csv("ai_student_impact_cleaned.csv")
df.head()
```

```python
df.info()
```

## 1. Generative AI Usage and Academic Performance

#### BQ1: Does using Gen AI more frequently relate to better or poorer academic performance?

```python
df[['Weekly_GenAI_Hours', 'Post_Semester_GPA']].corr()
```

```python
sns.set_style("darkgrid")
plt.figure(figsize = (10, 5))
sns.scatterplot(data = df, x = "Weekly_GenAI_Hours", y = "Post_Semester_GPA", color = "navy")
plt.title("Measured by Weekly GenAI Hours (0 to 40) versus Post-Semester GPA (1.0 to 4.0)", style = "italic")
plt.suptitle("Negligible Impact of Weekly GenAI Usage on Post-Semester GPA (r = -0.0186)", fontweight = "bold")
plt.xlabel("Weekly GenAI Hours", fontweight = "bold")
plt.ylabel("Post Semester GPA", fontweight = "bold")
plt.tight_layout()
plt.show()
```

#### BQ2: Which academic majors have the highest Gen AI usage, and how does their GPA compare?

```python
sns.set_style("darkgrid")
plt.figure(figsize = (10, 5))
sns.barplot(data = df, x = "Major_Category", y = "Post_Semester_GPA", color = "darkgreen")
plt.title("Average Post Semester GPA (0.0 to 3.5 Scale) Across Five Major Categories", style = "italic")
plt.suptitle("Uniform Academic Performance Across Majors with Average GPAs Near 3.35", fontweight = "bold")
plt.xlabel("Major Category", fontweight = "bold")
plt.ylabel("Average Post Semester GPA", fontweight = "bold")
plt.tight_layout()
plt.show()
```

```python
sns.set_style("darkgrid")
plt.figure(figsize = (10, 5))
sns.barplot(data = df, x = "Major_Category", y = "Weekly_GenAI_Hours", color = "skyblue")
plt.title("Average Weekly GenAI Hours (0 to 10+ Hours Scale) Across Five Major Categories", style = "italic")
plt.suptitle("STEM Leads Weekly GenAI Usage at ~10.5 Hours Compared to Other Academic Majors", fontweight = "bold")
plt.xlabel("Major Category", fontweight = "bold")
plt.ylabel("Average Weekly GenAI Hours", fontweight = "bold")
plt.tight_layout()
plt.show()
```

#### BQ3: Do students who use Gen AI have different purposes achieve different academic results?

```python
sns.set_style("darkgrid")
plt.figure(figsize = (10, 5))
sns.boxplot(data = df, x = "Primary_Use_Case", y = "Post_Semester_GPA", color = "crimson")
plt.title("Post Semester GPA (1.0 to 4.0 Scale) Across Five Primary Use Cases", style = "italic")
plt.suptitle("Similar Academic Performance Across GenAI Use Cases with Median GPAs Near 3.4 to 3.5", fontweight = "bold")
plt.xlabel("Primary Use Cases", fontweight = "bold")
plt.ylabel("Post Semester GPA", fontweight = "bold")
plt.tight_layout()
plt.show()
```

#### BQ4: Does using a wider variety of AI tools relate to higher post-semester GPA?

```python
sns.set_style("darkgrid")
plt.figure(figsize = (10, 5))
sns.barplot(data = df, x = "Tool_Diversity", y = "Post_Semester_GPA", color = "chocolate")
plt.title("Average Post Semester GPA (0.0 to 3.5 Scale) Across Distinct AI Tool Counts (1 to 5 Tools)", style = "italic")
plt.suptitle("Consistent Academic Performance Across AI Tool Counts with Average GPAs Near 3.35", fontweight = "bold")
plt.xlabel("Number of Distinct AI Tools", fontweight = "bold")
plt.ylabel("Average Post Semester GPA", fontweight = "bold")
plt.tight_layout()
plt.show()
```
